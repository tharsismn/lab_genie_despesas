# Laboratório: Genie Space - Despesas Públicas do Estado de São Paulo

## Objetivo

Este notebook é um laboratório para construir uma **Sala Genie** (Genie Space) que permitirá análises interativas sobre dados de **despesas públicas do Estado de São Paulo**.

## O que é uma Genie Space?

Genie Spaces são ambientes conversacionais do Databricks que permitem fazer perguntas em linguagem natural sobre seus dados. Com uma Genie Space configurada, usuários podem:

* Fazer perguntas sobre despesas públicas sem escrever SQL
* Obter insights instantâneos através de visualizações automáticas
* Explorar dados de forma intuitiva e interativa

## Dados

Trabalharemos com dados de despesas públicas do Estado de São Paulo, que incluem informações sobre:

* Órgãos e secretarias governamentais
* Categorias de despesas
* Valores empenhados, liquidados e pagos
* Períodos e datas das transações
* Fornecedores e beneficiários

## Próximos Passos

Neste laboratório, vamos:

1. Explorar e preparar os dados de despesas públicas
2. Criar e configurar tabelas no Unity Catalog
3. Configurar uma Genie Space otimizada para análise de despesas
4. Testar perguntas e validar respostas
